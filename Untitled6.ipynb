```python
# ============================================================
#                DNA SEQUENCE ANALYZER
# ============================================================

# ------------------------------------------------------------
# 1. VALIDATE DNA SEQUENCE
# ------------------------------------------------------------

def validate_sequence(sequence):

    sequence = sequence.upper().replace(" ", "")
    bases = "ATGC"
    invalid_bases = []

    if sequence == "":
        print("Sequence cannot be empty.")
        return False

    for base in sequence:
        if base not in bases and base not in invalid_bases:
            invalid_bases.append(base)

    if len(invalid_bases) == 0:
        print("Sequence is VALID.")
        return True

    else:
        print("Sequence is INVALID.")
        print("Invalid bases:", invalid_bases)
        return False


# ------------------------------------------------------------
# 2. SEQUENCE STATISTICS
# ------------------------------------------------------------

def sequence_statistics(sequence):

    sequence = sequence.upper().replace(" ", "")

    if len(sequence) == 0:
        print("Sequence cannot be empty.")
        return

    print("\n========== SEQUENCE STATISTICS ==========")

    print("Sequence:", sequence)
    print("Length:", len(sequence), "bp")

    print("\nNucleotide Count:")
    print("A =", sequence.count("A"))
    print("T =", sequence.count("T"))
    print("G =", sequence.count("G"))
    print("C =", sequence.count("C"))

    gc = sequence.count("G") + sequence.count("C")
    at = sequence.count("A") + sequence.count("T")

    gc_percentage = (gc / len(sequence)) * 100
    at_percentage = (at / len(sequence)) * 100

    print("\nGC Content:", round(gc_percentage, 2), "%")
    print("AT Content:", round(at_percentage, 2), "%")

    if gc_percentage > 50:
        print("Classification: GC-rich sequence")

    elif gc_percentage < 50:
        print("Classification: AT-rich sequence")

    else:
        print("Classification: Balanced GC/AT content")


# ------------------------------------------------------------
# 3. COMPLEMENT & REVERSE COMPLEMENT
# ------------------------------------------------------------

def complement_analysis(sequence):

    sequence = sequence.upper().replace(" ", "")

    if not validate_sequence(sequence):
        return

    complement = ""

    for base in sequence:

        if base == "A":
            complement += "T"

        elif base == "T":
            complement += "A"

        elif base == "G":
            complement += "C"

        elif base == "C":
            complement += "G"

    reverse = sequence[::-1]
    reverse_complement = complement[::-1]

    print("\n========== COMPLEMENT ANALYSIS ==========")

    print("Original Sequence      :", sequence)
    print("Complement             :", complement)
    print("Reverse Sequence       :", reverse)
    print("Reverse Complement     :", reverse_complement)


# ------------------------------------------------------------
# 4. MOTIF SEARCH
# ------------------------------------------------------------

def motif_search(sequence):

    sequence = sequence.upper().replace(" ", "")

    motif = input("Enter the motif to search: ").upper().replace(" ", "")

    if motif == "":
        print("Motif cannot be empty.")
        return

    positions = []
    start = 0

    while True:

        position = sequence.find(motif, start)

        if position == -1:
            break

        positions.append(position + 1)

        start = position + 1

    print("\n========== MOTIF SEARCH ==========")

    print("Motif:", motif)

    if len(positions) == 0:

        print("Motif not found.")

    else:

        print("Motif found.")
        print("Number of occurrences:", len(positions))
        print("Positions:", positions)


# ------------------------------------------------------------
# 5. MUTATION ANALYSIS
# ------------------------------------------------------------

def mutation_analysis(reference, sample):

    reference = reference.upper().replace(" ", "")
    sample = sample.upper().replace(" ", "")

    print("\n========== MUTATION ANALYSIS ==========")

    if len(reference) != len(sample):

        print("Reference and sample sequences have different lengths.")
        print("Position-by-position comparison cannot be performed.")
        return

    mutation_count = 0
    transitions = 0
    transversions = 0

    print("\nMutations:")

    for i in range(len(reference)):

        ref_base = reference[i]
        sample_base = sample[i]

        if ref_base != sample_base:

            mutation_count += 1

            # Transition
            if ((ref_base == "A" and sample_base == "G") or
                (ref_base == "G" and sample_base == "A") or
                (ref_base == "C" and sample_base == "T") or
                (ref_base == "T" and sample_base == "C")):

                mutation_type = "Transition"
                transitions += 1

            else:

                mutation_type = "Transversion"
                transversions += 1

            print(
                "Position", i + 1, ":",
                ref_base, "->", sample_base,
                "(", mutation_type, ")"
            )

    mutation_percentage = (
        mutation_count / len(reference)
    ) * 100

    print("\nTotal Mutations:", mutation_count)
    print("Transitions:", transitions)
    print("Transversions:", transversions)

    print(
        "Mutation Percentage:",
        round(mutation_percentage, 2),
        "%"
    )


# ------------------------------------------------------------
# 6. SEQUENCE SIMILARITY ANALYSIS
# ------------------------------------------------------------

# ------------------------------------------------------------
# 6. SEQUENCE SIMILARITY ANALYSIS
# ------------------------------------------------------------

def similarity_analysis(reference, sample):

    reference = reference.upper().replace(" ", "")
    sample = sample.upper().replace(" ", "")

    print("\n========== SEQUENCE SIMILARITY ==========")

    if len(reference) == 0 or len(sample) == 0:
        print("Sequences cannot be empty.")
        return

    # Find the minimum sequence length
    comparison_length = min(len(reference), len(sample))

    matches = 0
    mismatches = 0

    for i in range(comparison_length):

        if reference[i] == sample[i]:
            matches += 1
        else:
            mismatches += 1

    similarity = (matches / comparison_length) * 100

    print("Reference Length:", len(reference))
    print("Sample Length:", len(sample))
    print("Comparison Length:", comparison_length)

    print("Matching Positions:", matches)
    print("Mismatching Positions:", mismatches)

    print(
        "Sequence Similarity:",
        round(similarity, 2),
        "%"
    )

    if len(reference) != len(sample):
        print("\nNote: Only the first",
              comparison_length,
              "positions were compared.")




# ============================================================
#                     MAIN PROGRAM
# ============================================================

def main():

    print("=" * 55)
    print("              DNA SEQUENCE ANALYZER")
    print("=" * 55)

    reference = input(
        "\nEnter Reference DNA Sequence: "
    ).upper().replace(" ", "")

    sample = input(
        "Enter Sample DNA Sequence: "
    ).upper().replace(" ", "")

    while True:

        print("\n")
        print("=" * 55)
        print("                    MAIN MENU")
        print("=" * 55)

        print("1. Validate DNA Sequences")
        print("2. Sequence Statistics")
        print("3. Complement & Reverse Complement")
        print("4. Motif Search")
        print("5. Mutation Analysis")
        print("6. Sequence Similarity Analysis")
        print("7. Exit")

        print("=" * 55)

        choice = input("Enter your choice (1-7): ")

        # Option 1
        if choice == "1":

            print("\n--- REFERENCE SEQUENCE ---")
            validate_sequence(reference)

            print("\n--- SAMPLE SEQUENCE ---")
            validate_sequence(sample)

        # Option 2
        elif choice == "2":

            print("\n--- REFERENCE SEQUENCE ---")
            sequence_statistics(reference)

            print("\n--- SAMPLE SEQUENCE ---")
            sequence_statistics(sample)

        # Option 3
        elif choice == "3":

            print("\n--- REFERENCE SEQUENCE ---")
            complement_analysis(reference)

            print("\n--- SAMPLE SEQUENCE ---")
            complement_analysis(sample)

        # Option 4
        elif choice == "4":

            print("\n--- REFERENCE SEQUENCE ---")
            motif_search(reference)

            print("\n--- SAMPLE SEQUENCE ---")
            motif_search(sample)

        # Option 5
        elif choice == "5":

            mutation_analysis(reference, sample)

        # Option 6
        elif choice == "6":

            similarity_analysis(reference, sample)

        # Exit
        elif choice == "7":

            print("\nThank you for using DNA Sequence Analyzer!")
            print("Analysis completed.")
            break

        else:

            print("\nInvalid choice.")
            print("Please enter a number from 1 to 7.")


# ============================================================
#                   START PROGRAM
# ============================================================

main()
```
